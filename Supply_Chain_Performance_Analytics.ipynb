```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/dataset.csv', encoding='latin-1')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Order Zipcode', 'Product Description'])

df.head()
```

```python
df.isnull().sum()
```

```python
# Count missing values per column
missing_counts = df.isnull().sum()

# Select columns with <= 3 missing values
cols_to_clean = missing_counts[(missing_counts >= 1) & (missing_counts <= 8)].index

print(cols_to_clean)
```

```python
df_cleaned = df.dropna(subset=cols_to_clean)
```

```python
df.isnull().sum()
```

```python
duplicate_rows = df.duplicated().sum()
display(duplicate_rows)
```

```python
df.rename(columns={'order date (DateOrders)':'order_date', 'shipping date (DateOrders)': 'shipping_date'}, inplace=True)
df.columns = [col.lower().replace(' ', '_') for col in df.columns]

```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['shipping_date'] = pd.to_datetime(df['shipping_date'])
```

```python
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_quarter'] = df['order_date'].dt.quarter
df['order_dayofweek'] = df['order_date'].dt.dayofweek
df.head()
```

```python
df['ddi'] = df['days_for_shipping_(real)'] - df['days_for_shipment_(scheduled)']
df.head()
```

```python
df['is_high_risk'] = (df['late_delivery_risk'] == 1) & (df['ddi'] > 0)
```

```python
late_risk_by_shipping_mode = df.groupby('shipping_mode')['late_delivery_risk'].mean().sort_values(ascending=False)
global_mean_late_rate = df['late_delivery_risk'].mean()

plt.figure(figsize=(12, 7))
sns.barplot(x=late_risk_by_shipping_mode.values, y=late_risk_by_shipping_mode.index, palette='viridis', hue=late_risk_by_shipping_mode.index, legend=False)

# Add percentage labels to bars
for index, value in enumerate(late_risk_by_shipping_mode.values):
    plt.text(value + 0.01, index, f'{value:.1%}', va='center')

# Add global mean line
plt.axvline(x=global_mean_late_rate, color='red', linestyle='--', linewidth=2, label=f'Global Mean ({global_mean_late_rate:.1%})')

plt.title('Late Delivery Risk Rate by Shipping Mode', fontsize=16)
plt.xlabel('Late Delivery Risk Rate', fontsize=12)
plt.ylabel('Shipping Mode', fontsize=12)
plt.xlim(0, 1) # Set x-axis limit from 0 to 1 for percentage
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
median_ddi_by_shipping_mode = df.groupby('shipping_mode')['ddi'].median().sort_values(ascending=False)
ordered_shipping_modes = median_ddi_by_shipping_mode.index

data_to_plot = [df[df['shipping_mode'] == mode]['ddi'].dropna() for mode in ordered_shipping_modes]

plt.figure(figsize=(12, 8))
plt.boxplot(data_to_plot,labels=ordered_shipping_modes, vert=False)
plt.title('Distribution of DDI by Shipping Mode Ordered by Median DDI', fontsize=16)
plt.xlabel('Delivery Delay Index (DDI)', fontsize=12)
plt.ylabel('Shipping Mode', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
late_risk_by_market = df.groupby('market')['late_delivery_risk'].mean().sort_values(ascending=False)
global_mean_late_rate = df['late_delivery_risk'].mean()

plt.figure(figsize=(12, 7))
sns.barplot(x=late_risk_by_market.values, y=late_risk_by_market.index, palette='viridis', hue=late_risk_by_market.index, legend=False)

# Add percentage labels to bars
for index, value in enumerate(late_risk_by_market.values):
    plt.text(value + 0.01, index, f'{value:.1%}', va='center')

# Add global mean line
plt.axvline(x=global_mean_late_rate, color='red', linestyle='--', linewidth=2, label=f'Global Mean ({global_mean_late_rate:.1%})')

plt.title('Late Delivery Risk Rate by Market', fontsize=16)
plt.xlabel('Late Delivery Risk Rate', fontsize=12)
plt.ylabel('Market', fontsize=12)
plt.xlim(0, 1) # Set x-axis limit from 0 to 1 for percentage
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
region_analysis = df.groupby('order_region').agg(
    late_rate=('late_delivery_risk', 'mean'),
    avg_delay=('ddi', 'mean'),
    total_orders=('order_id', 'count')
).sort_values(by='late_rate', ascending=False)

print(region_analysis)
```

```python
monthly_late_deliveries_count = df.groupby('order_month')['late_delivery_risk'].sum().sort_index()

plt.figure(figsize=(12, 7))
sns.lineplot(x=monthly_late_deliveries_count.index, y=monthly_late_deliveries_count.values, marker='o', color='blue')

plt.title('Monthly Late Delivery Count Trend', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Count of Late Deliveries', fontsize=12)
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(axis='both', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
revenue_impact_mean = df.groupby('late_delivery_risk')['sales_per_customer'].mean()
print('Average Revenue Lost per Order by Late Delivery Risk:\n', revenue_impact_mean)

total_revenue_lost = df[df['late_delivery_risk'] == 1]['sales_per_customer'].sum()
print('\nTotal Revenue at risk Due to Late Deliveries:', total_revenue_lost)
```

```python
category_analysis = df.groupby('category_name').agg(
    late_delivery_mean=('late_delivery_risk', 'mean'),
    late_delivery_count=('late_delivery_risk', 'sum'),
    avg_order_value=('sales_per_customer', 'mean')
)

category_analysis['revenue_at_risk'] = category_analysis['late_delivery_count'] * category_analysis['avg_order_value']

category_analysis = category_analysis.sort_values(by='revenue_at_risk', ascending=False)

print(category_analysis.head())
```

```python
category_late_rate = df.groupby('category_name')['late_delivery_risk'].mean().sort_values(ascending=False).head(10)

global_mean_late_rate = df['late_delivery_risk'].mean()
plt.figure(figsize=(12, 7))
sns.barplot(x=category_late_rate.values, y=category_late_rate.index, palette='viridis', hue=category_late_rate.index, legend=False)

plt.axvline(x=global_mean_late_rate, color='red', linestyle='--', linewidth=2, label=f'Global Mean ({global_mean_late_rate:.1%})')

# Add percentage labels to bars
for index, value in enumerate(category_late_rate.values):
    plt.text(value + 0.01, index, f'{value:.1%}', va='center')

plt.title('Top 10 Product Categories by Late Delivery Risk Rate', fontsize=16)
plt.xlabel('Late Delivery Risk Rate', fontsize=12)
plt.ylabel('Product Category', fontsize=12)
plt.xlim(0, 1) # Set x-axis limit from 0 to 1 for percentage
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
segment_analysis = df.groupby('customer_segment').agg(
    late_rate=('late_delivery_risk', 'mean'),
    avg_delay=('ddi', 'mean'),
    avg_profit=('benefit_per_order', 'mean')
).reset_index()
display(segment_analysis)
```

```python
df['route'] = df['order_region'] + ' | ' + df['shipping_mode']

route_analysis = df.groupby('route').agg(
    late_rate=('late_delivery_risk', 'mean'),
    total_orders=('order_id', 'count')
).query('total_orders > 500').sort_values(by='late_rate', ascending=False)

print("\nTop 10 Worst Performing Routes by Late Delivery Rate (filtered by >500 orders):\n")
print(route_analysis.head(10))
```

```python
status_to_plot = ['COMPLETE', 'PENDING', 'CLOSED', 'CANCELED']

# Filter for specified statuses and group by market and order_status
market_status_counts = df[df['order_status'].isin(status_to_plot)].groupby(['market', 'order_status']).size().unstack(fill_value=0)

# Plotting the stacked bar chart
market_status_counts.plot(kind='bar', stacked=True, figsize=(14, 8), cmap='viridis')
plt.title('Order Status Distribution by Market', fontsize=16)
plt.xlabel('Market', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Order Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
payment_order_status = pd.crosstab(df['type'], df['order_status'])

print("Cross-tabulation of Payment Type and Order Status:")
display(payment_order_status)
```

```python
df['order_type'] = df['order_item_quantity'].apply(lambda x: 'Single Item' if x == 1 else 'Bulk/Multi-Item')

order_type_analysis = df.groupby('order_type').agg(
    late_delivery_risk=('late_delivery_risk', 'mean'),
    mean_ddi=('ddi', 'mean')
).reset_index()

print("\nLate Delivery Risk and Median DDI by Order Type:")
display(order_type_analysis)
```

```python
department_analysis = df.groupby('department_name').agg(
    late_delivery_risk=('late_delivery_risk', 'mean')
).sort_values(by='late_delivery_risk', ascending=False)

display(department_analysis)
```
